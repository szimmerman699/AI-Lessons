### Step 5 — Understand the dual-path pattern

Run the cell below. Read the code first — it shows the pattern every exercise notebook in this course uses to load secrets. The same notebook works in Colab (via Colab Secrets) and locally (via environment variables).

````python
import os

def load_secret(name: str) -> str:
    """Load a secret from Colab Secrets or environment variables."""
    try:
        from google.colab import userdata
        value = userdata.get(name)
        source = 'Colab Secrets'
    except Exception:
        value = os.environ.get(name)
        source = 'environment variable'

    if not value:
        raise ValueError(
            f'{name} not found. '
            f'In Colab: add it via the key icon in the sidebar. '
            f'Locally: export {name}=... before launching Jupyter.'
        )

    print(f'Loaded {name} from {source}')
    return value

key = load_secret('ANTHROPIC_API_KEY')
print(f'Key starts with: {key[:7]}...')
````

### Step 7 — Register the kernel

```bash

```

A **kernel** is the Python process that runs behind a notebook. Each notebook connects to one kernel. When you "Restart kernel," you kill that process and start a fresh one — all variables are gone. This is the fix for hidden-state bugs.

### Step 8 — Launch Jupyter Lab

```bash
jupyter lab
```

Your browser opens to `http://localhost:8888/lab`. If port 8888 is taken, use `jupyter lab --port 8889`.


### Step 10 — Explore kernel controls

In Jupyter Lab, find the **Kernel** menu:

- **Restart Kernel:** Kills the Python process, clears all variables. Use this when you suspect hidden state.
- **Restart Kernel and Run All Cells:** The gold standard — proves the notebook runs top-to-bottom from a clean state.
- **Interrupt Kernel:** Stops a running cell (like Ctrl+C in a terminal).

**Try it:** Run "Restart Kernel and Run All Cells" now. If it fails, fix the failing cell. A notebook that doesn't run top-to-bottom is broken.

## Your turn

Add a new code cell below. Write a function called `notebook_metadata()` that returns a dictionary with:

- `"lesson"`: the lesson name
- `"runtime"`: whether you're in Colab or local Jupyter (detect it programmatically)
- `"python_version"`: the Python version
- `"key_source"`: where the API key was loaded from (Colab Secrets or env var)

Print the result. This is a small exercise in writing utility code inside a notebook — the kind of cell you'll reuse and adapt throughout the course.

````python
import sys
import os


def notebook_metadata():
    """
    Returns a dictionary with notebook metadata.
    
    Returns:
        dict: Contains 'lesson', 'runtime', 'python_version', 'key_source'
    """
    
    # Detect runtime
    try:
        from google.colab import userdata
        runtime = "Colab"
    except ImportError:
        runtime = "Local Jupyter"
    
    # Python version
    python_version = f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}"
    
    # Detect API key source
    key_source = "Not Found"
    
    if runtime == "Colab":
        try:
            from google.colab import userdata
            for key_name in ['ANTHROPIC_API_KEY', 'OPENAI_API_KEY', 'API_KEY']:
                try:
                    userdata.get(key_name)
                    key_source = f"Colab Secrets ({key_name})"
                    break
                except userdata.SecretNotFoundError:
                    continue
        except ImportError:
            pass
    
    # Check environment variables
    if key_source == "Not Found":
        for key_name in ['ANTHROPIC_API_KEY', 'OPENAI_API_KEY', 'API_KEY']:
            if key_name in os.environ:
                key_source = f"Environment Variable ({key_name})"
                break
    
    # Get lesson name
    lesson = "Unknown"
    if runtime == "Colab":
        try:
            from google.colab import _ipython_utils
            notebook_name = _ipython_utils.get_notebook_name()
            lesson = notebook_name if notebook_name else "Unknown"
        except (ImportError, AttributeError):
            pass
    
    return {
        "lesson": lesson,
        "runtime": runtime,
        "python_version": python_version,
        "key_source": key_source
    }


# Print the results
metadata = notebook_metadata()

print("=" * 50)
print("NOTEBOOK METADATA")
print("=" * 50)
for key, value in metadata.items():
    print(f"{key:20s}: {value}")
print("=" * 50)

````
